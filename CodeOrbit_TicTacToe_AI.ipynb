# CodeOrbit Tech Internship — Project 2: Tic-Tac-Toe with Simple AI\n\nYou are **X** and the computer is **O**. The AI uses Minimax. Run the code cell and click the board.

```python
import math
import ipywidgets as widgets
from IPython.display import display

# X is the human player; O is the computer.
board = [""] * 9
human, computer = "X", "O"
game_over = False

def check_winner(state):
    """Return X/O for a winner, Draw for a full board, or None."""
    lines = [(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]
    for a,b,c in lines:
        if state[a] and state[a] == state[b] == state[c]:
            return state[a]
    return "Draw" if "" not in state else None

def minimax(state, maximizing):
    """Evaluate possible moves: computer win=1, human win=-1, draw=0."""
    result = check_winner(state)
    if result == computer: return 1
    if result == human: return -1
    if result == "Draw": return 0

    if maximizing:
        best = -math.inf
        for i in range(9):
            if state[i] == "":
                state[i] = computer
                best = max(best, minimax(state, False))
                state[i] = ""
        return best
    else:
        best = math.inf
        for i in range(9):
            if state[i] == "":
                state[i] = human
                best = min(best, minimax(state, True))
                state[i] = ""
        return best

def computer_move():
    """Pick the highest-scoring move with Minimax."""
    best_score, best_move = -math.inf, None
    for i in range(9):
        if board[i] == "":
            board[i] = computer
            score = minimax(board, False)
            board[i] = ""
            if score > best_score:
                best_score, best_move = score, i
    if best_move is not None:
        board[best_move] = computer

def refresh():
    for i, button in enumerate(buttons):
        button.description = board[i] if board[i] else " "
        button.disabled = bool(board[i]) or game_over
        button.style.button_color = "#D9EAF7" if board[i] == human else ("#DDF3E4" if board[i] == computer else "#F3F4F6")
    result = check_winner(board)
    if result == human:
        status.value = "<b style='color:green'>You win! 🎉</b>"
    elif result == computer:
        status.value = "<b style='color:#b42318'>Computer wins!</b>"
    elif result == "Draw":
        status.value = "<b>It's a draw!</b>"
    elif not game_over:
        status.value = "<b>Your turn (X)</b>"

def player_move(index):
    global game_over
    if game_over or board[index] != "":
        return
    board[index] = human
    result = check_winner(board)
    if result is not None:
        game_over = True
        refresh()
        return
    computer_move()
    if check_winner(board) is not None:
        game_over = True
    refresh()

buttons = []
for idx in range(9):
    b = widgets.Button(description=" ", layout=widgets.Layout(width="85px", height="85px"),
                       style={"font_weight":"bold", "font_size":"24px"})
    b.on_click(lambda btn, i=idx: player_move(i))
    buttons.append(b)

status = widgets.HTML("<b>Your turn (X)</b>")
restart = widgets.Button(description="Restart Game", button_style="info", icon="refresh")

def restart_game(_):
    global board, game_over
    board = [""] * 9
    game_over = False
    refresh()

restart.on_click(restart_game)
display(widgets.VBox([
    widgets.HTML("<h2 style='color:#12344D'>CodeOrbit Tech — Tic-Tac-Toe AI</h2><p>You are X; computer is O. AI uses Minimax.</p>"),
    status,
    widgets.HBox(buttons[0:3]), widgets.HBox(buttons[3:6]), widgets.HBox(buttons[6:9]),
    restart
]))
refresh()
```
